## 1. Introduction to NumPy

NumPy stands for **Numerical Python**. It is the core library for scientific and numerical computing in Python.

### Why NumPy is Essential in AI/ML:
- **Multi-dimensional Arrays:** Provides the `ndarray` object representing vectors (1D) and matrices (2D/3D).
- **Vectorized Operations:** Performs arithmetic and mathematical operations without slow Python `for` loops.
- **Memory Efficiency:** Stores data in contiguous memory blocks, making computations much faster.
- **Foundation of AI/ML:** Libraries like Pandas, Scikit-learn, TensorFlow, and PyTorch are built on top of NumPy arrays.

## 2. Importing NumPy & Version Check

```python
import numpy as np

print("NumPy successfully imported. Version:", np.__version__)
```

## 3. Creating a 1D Array
A 1D array represents a single vector of numbers.

```python
# Create 1D array from a Python list
arr_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr_1d)
print("Shape:", arr_1d.shape)
print("Dimensions:", arr_1d.ndim)
print("Data Type:", arr_1d.dtype)
```

## 4. Creating a 2D Array
A 2D array represents a grid or table of rows and columns (matrix).

```python
# Create a 2D array (matrix)
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("2D Array:\n", arr_2d)
print("Shape:", arr_2d.shape, "(2 rows, 3 columns)")
print("Dimensions:", arr_2d.ndim)
print("Total Elements:", arr_2d.size)
```

## 5. Indexing in Arrays
Accessing individual elements using 0-based index positions.

```python
# 1D array indexing
print("First element of 1D array:", arr_1d[0])
print("Last element of 1D array:", arr_1d[-1])

# 2D array indexing [row, column]
print("Element at Row 0, Column 1 in 2D array:", arr_2d[0, 1])
print("Element at Row 1, Column 2 in 2D array:", arr_2d[1, 2])
```

## 6. Slicing in Arrays
Extracting sub-arrays using `[start:stop:step]` syntax.

```python
# 1D slicing
print("1D slice (index 1 to 3):", arr_1d[1:4])
print("1D slice (first 3 elements):", arr_1d[:3])

# 2D slicing [row_slice, col_slice]
print("2D slice (first row):", arr_2d[0, :])
print("2D slice (all rows, column 1 to end):\n", arr_2d[:, 1:])
```

## 7. Basic Mathematical Operations
Using built-in statistical and mathematical functions.

```python
numbers = np.array([10, 25, 40, 55, 70])

print("Numbers:", numbers)
print("Sum    :", np.sum(numbers))
print("Mean   :", np.mean(numbers))
print("Max    :", np.max(numbers))
print("Min    :", np.min(numbers))

# Element-wise arithmetic
print("Add 5 to all elements:", numbers + 5)
print("Multiply all by 2    :", numbers * 2)
```

## 8. Python List vs. NumPy Array Comparison

| Feature | Python List | NumPy Array |
| :--- | :--- | :--- |
| **Data Types** | Heterogeneous (can mix strings, ints, objects) | Homogeneous (single numeric data type) |
| **Memory** | Non-contiguous memory allocation | Contiguous memory allocation (compact) |
| **Arithmetic** | Concatenates or repeats (`+`, `*`) | Element-wise vectorized calculation |
| **Performance** | Slower for large numerical calculations | Optimized in C, highly efficient for AI/ML |

```python
# Demonstration of '+' behavior
py_list = [1, 2, 3]
np_arr = np.array([1, 2, 3])

print("List '+' operation (concatenation):", py_list + py_list)
print("NumPy '+' operation (element-wise):", np_arr + np_arr)
```

## 9. Student Marks Analysis Using NumPy

Here we analyze a dataset of marks for **4 students** across **3 subjects** (Subject 1, Subject 2, Subject 3) using basic NumPy operations.

```python
# 2D array: Rows = Students (1 to 4), Columns = Subjects (1 to 3)
marks = np.array([
    [85, 78, 90],  # Student 1
    [92, 88, 95],  # Student 2
    [70, 65, 80],  # Student 3
    [88, 82, 84]   # Student 4
])

print("Student Marks Matrix (Rows = Students, Cols = Subjects):\n", marks)

# 1. Overall Statistics
print("\n--- Overall Class Metrics ---")
print("Total Marks Scored across all:", np.sum(marks))
print(f"Overall Class Average        : {np.mean(marks):.2f}")
print("Highest Score in Class       :", np.max(marks))
print("Lowest Score in Class        :", np.min(marks))

# 2. Per-Student Metrics (sum across columns, axis=1)
student_totals = np.sum(marks, axis=1)
student_averages = np.mean(marks, axis=1)

print("\n--- Per-Student Performance ---")
for i in range(len(student_totals)):
    print(f"Student {i+1}: Total = {student_totals[i]}, Average = {student_averages[i]:.2f}")

# 3. Per-Subject Average (mean down rows, axis=0)
subject_averages = np.mean(marks, axis=0)
print("\n--- Per-Subject Averages ---")
for j in range(len(subject_averages)):
    print(f"Subject {j+1} Average: {subject_averages[j]:.2f}")
```
