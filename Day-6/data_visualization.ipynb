## 1. Introduction to Data Visualization

**Data Visualization** is the process of representing data graphically so that patterns, distributions, and comparisons can be understood more easily.

- **Matplotlib:** A comprehensive Python library used to create static, animated, and interactive visualizations.
- **Seaborn:** A Python visualization library built on top of Matplotlib that provides convenient, high-level functions for statistical graphics.

### Charts Practiced Today:
- **Bar Chart:** Used to compare discrete values across distinct categories.
- **Line Chart:** Used to display values across an ordered sequence and observe changes.
- **Histogram:** Used to understand the frequency distribution of numerical values.
- **Pie Chart:** Used to show proportional contributions of categories to a whole.

## 2. Import Libraries
Importing Pandas, Matplotlib, and Seaborn with verified versions.

```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

print("Pandas Version    :", pd.__version__)
print("Matplotlib Version:", matplotlib.__version__)
print("Seaborn Version   :", sns.__version__)
```

## 3. Load Day 5 Dataset Dynamically
Reading the single source dataset directly from `../Day-5/dataset.csv` without duplicating files.

```python
df = pd.read_csv("../Day-5/dataset.csv")
print("Day 5 Dataset loaded dynamically!")
print(f"Shape: {df.shape} ({df.shape[0]} rows, {df.shape[1]} columns)")
```

## 4. Basic Dataset Check
Previewing the first 5 records of the dataset.

```python
df.head()
```

## 5. Bar Chart: Employees by Department
Comparing employee counts across departments using `plt.bar()`.

```python
# Bar Chart
dept_counts = df["Department"].value_counts()

plt.figure(figsize=(7, 4))
plt.bar(dept_counts.index, dept_counts.values, color="skyblue", edgecolor="black")
plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

## 6. Line Chart: Employee Salary
Displaying employee salary across employee IDs using `plt.plot()`.

```python
# Line Chart
salary_df = df.dropna(subset=["Salary"]).copy()

plt.figure(figsize=(8, 4))
plt.plot(salary_df["Employee_ID"].astype(str), salary_df["Salary"], marker="o", color="green", linewidth=2)
plt.title("Employee Salary across Employee IDs")
plt.xlabel("Employee ID")
plt.ylabel("Salary")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 7. Histogram: Age Distribution
Visualizing the numerical distribution of employee ages using `plt.hist()`.

```python
# Histogram
plt.figure(figsize=(7, 4))
plt.hist(df["Age"], bins=5, color="coral", edgecolor="black")
plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Frequency (Number of Employees)")
plt.tight_layout()
plt.show()
```

## 8. Pie Chart: Department Distribution
Displaying the proportion of employees across departments using `plt.pie()`.

```python
# Pie Chart
plt.figure(figsize=(6, 6))
plt.pie(dept_counts.values, labels=dept_counts.index, autopct="%1.1f%%", startangle=140, colors=["#66b3ff", "#99ff99", "#ffcc99", "#ff9999"])
plt.title("Department Distribution")
plt.tight_layout()
plt.show()
```
