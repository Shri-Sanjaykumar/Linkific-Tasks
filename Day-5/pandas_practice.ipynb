## 1. Introduction to Pandas

**Pandas** is the primary data manipulation and analysis library in Python. It provides fast, flexible data structures—most notably the **DataFrame**—designed to make working with structured (tabular, CSV, SQL) data intuitive and efficient for machine learning workflows.

## 2. Importing Pandas & Version Verification

```python
import pandas as pd

print("Pandas successfully imported. Version:", pd.__version__)
```

## 3. Loading a CSV Dataset
Using `pd.read_csv()` to load our tabular dataset (`dataset.csv`) into a Pandas DataFrame.

```python
df = pd.read_csv("dataset.csv")
print("Dataset loaded successfully!")
```

## 4. Understanding the DataFrame
A **DataFrame** is a two-dimensional labeled data structure with columns of potentially different types (like a spreadsheet or SQL table).

```python
print("Data Structure Type:", type(df))
print("Columns:", df.columns)
```

## 5. Displaying First 5 Rows
Using `df.head()` to inspect the top 5 records.

```python
df.head()
```

## 6. Displaying Last 5 Rows
Using `df.tail()` to inspect the bottom 5 records.

```python
df.tail()
```

## 7. Checking Rows and Columns
Using `df.shape` to check the dataset dimensions `(rows, columns)`.

```python
print("Dataset Shape:", df.shape)
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

## 8. Checking Missing Values
Using `df.isnull().sum()` to count missing (null/NaN) entries per column.

```python
print("Missing values per column:")
print(df.isnull().sum())
```

## 9. Filtering Data Using Conditions
Filtering records based on conditional logic.

```python
# Condition 1: High earners (Salary > 70,000)
print("--- Condition 1: Employees with Salary > 70,000 ---")
print(df[df["Salary"] > 70000])

# Condition 2: Specific department (Department == 'Data Science')
print("\n--- Condition 2: Employees in Data Science Department ---")
print(df[df["Department"] == "Data Science"])
```

## 10. Summary Statistics
Using `df.describe()` to compute statistical metrics (count, mean, std, min, 25%, 50%, 75%, max) for numeric columns.

```python
df.describe()
```
